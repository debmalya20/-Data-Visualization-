## BASIC OF MATPLOTLIB

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
x=np.linspace(1,5,10)
```

```python
x
y=x**4
```

```python
y
```

```python
plt.plot(x,y) # we use this for single line graph
plt.title("Hello") # this allowes us to give the graph name
plt.xlabel("lebel X") # this is to change the name of the lebel X
plt.ylabel("Label y") # this is to change the name of the lebel Y
```

## SUBPLOT IN MATPLOTLIB

```python
#SUBPLOT IS A FUNCTION THAT HELPS US TO CREATE MORE THAN 1 PLOT(graphs) INSIDE A BIG BLANK SQURE SPACE
```

```python
plt.subplot(2,2,1)
plt.subplot(2,2,2)
plt.subplot(2,2,3)
plt.subplot(2,2,4) #so this is how we can make subplots
plt.plot(x,y) # this is how we can make plot inside a subplot.by dinh the plot down of the subplot
```

## OBJECT ORIANTED WAY
