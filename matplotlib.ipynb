## BASIC OF MATPLOTLIB

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
x=np.linspace(1,5,10)
```

```python
x
y=x**4
```

```python
y
```

```python
plt.plot(x,y) # we use this for single line graph
plt.title("Hello") # this allowes us to give the graph name
plt.xlabel("lebel X") # this is to change the name of the lebel X
plt.ylabel("Label y") # this is to change the name of the lebel Y
```

## SUBPLOT IN MATPLOTLIB

```python
#SUBPLOT IS A FUNCTION THAT HELPS US TO CREATE MORE THAN 1 PLOT(graphs) INSIDE A BIG BLANK SQURE SPACE
```

```python
plt.subplot(2,2,1)
plt.subplot(2,2,2)
plt.subplot(2,2,3)
plt.subplot(2,2,4) #so this is how we can make subplots
plt.plot(x,y) # this is how we can make plot inside a subplot.by dinh the plot down of the subplot
```

## OBJECT ORIANTED WAY

```python
fig=plt.figure(figsize=(7,5),dpi=100) #figisze helps us to custom out own size of canvas or white sheet. and dpi helps us to make the frame sharp.
axes1=fig.add_axes([0.3,0.3,0.8,0.8]) # this is how can create a canvas of my own choice
axes1.plot(x,y) 
axes2=fig.add_axes([0.3,0.8,0.3,0.3]) # this is how we can add a new plot in the existing plot, and by chmaging the ration we cna chmagr the size
axes2.plot(x,y) 
```
