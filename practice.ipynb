```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
x=np.linspace(1,5,10)
x
y=x**2
y
```

```python
plt.plot(x,y)
plt.title("Graph")
plt.xlabel("This is x lable")
plt.ylabel("This is y lable")
```

```python
plt.subplot(2,2,1)
plt.plot(x,y)
plt.subplot(2,2,2)
plt.subplot(2,2,3)
plt.subplot(2,2,4)
```

```python
fig1=plt.figure(figsize=(7,5))
canvas=fig1.add_axes([0.3,0.3,0.4,0.4])
canvas.plot(x,y)
canvas2=fig1.add_axes([0.2,0.52,0.2,0.2])
canvas2.plot(x,y)
```

```python
plt.scatter(x,y)
```

```python
plt.hist(x,y)
```
